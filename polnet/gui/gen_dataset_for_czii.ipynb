```python
output_directory = "/home/tomo/kaggle/polnet/output/241130"
os.makedirs(output_directory, exist_ok=True)
```

```python
use_pns_list = [
    "/home/tomo/kaggle/polnet/data/in_10A/1fa2_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/6cvm_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/6z6u_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/8q3f_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/6utj_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/5mrc_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/4v4r_50S_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/4cr2_10A.pns",
    "/home/tomo/kaggle/polnet/data/in_10A/2uv8_10A.pns",
]

use_mbs_list = [
    "/home/tomo/kaggle/polnet/data/in_mbs/ellipse.mbs",
    "/home/tomo/kaggle/polnet/data/in_mbs/sphere.mbs",
    "/home/tomo/kaggle/polnet/data/in_mbs/toroid.mbs",
]

use_hns_list = [
    "/home/tomo/kaggle/polnet/data/in_helix/actin.hns",
    "/home/tomo/kaggle/polnet/data/in_helix/mt.hns",
]
```

```python
# fixed parameters
ntomos = 100
voi_shape = (630, 630, 200)
x = voi_shape[0]; y = voi_shape[1]; z = voi_shape[2]
voi_off = ((int(x * 0.025), int(x * 0.975)), 
            (int(y * 0.025), int(y * 0.975)), 
            (int(z * 0.025), int(z * 0.975)) )
voi_size = 10
mmer_tries = 20
pmer_tries = 100
surf_dec = 0.9
tilt_angs = range(-45, 46, 3)
detector_snr = [0.2, 3.0]
malign_mn = 1
malign_mx = 5
malign_sg = 0.5
```

```python
# 使用リスト、パラメータをファイル保存
with open(output_directory + "/params.txt", "w") as f:
    f.write("use_pns_list = " + str(use_pns_list) + "\n")
    f.write("use_mbs_list = " + str(use_mbs_list) + "\n")
    f.write("use_hns_list = " + str(use_hns_list) + "\n")
    f.write("ntomos = " + str(ntomos) + "\n")
    f.write("voi_shape = " + str(voi_shape) + "\n")
    f.write("voi_off = " + str(voi_off) + "\n")
    f.write("voi_size = " + str(voi_size) + "\n")
    f.write("mmer_tries = " + str(mmer_tries) + "\n")
    f.write("pmer_tries = " + str(pmer_tries) + "\n")
    f.write("surf_dec = " + str(surf_dec) + "\n")
    f.write("tilt_angs = " + str(tilt_angs) + "\n")
    f.write("detector_snr = " + str(detector_snr) + "\n")
    f.write("malign_mn = " + str(malign_mn) + "\n")
    f.write("malign_mx = " + str(malign_mx) + "\n")
    f.write("malign_sg = " + str(malign_sg) + "\n")
```

```python
all_features2(ntomos, voi_shape, output_directory, voi_off, voi_size, mmer_tries, pmer_tries,
                use_mbs_list, use_hns_list, use_pns_list, [], surf_dec, tilt_angs,
                detector_snr, malign_mn, malign_mx, malign_sg)
```
