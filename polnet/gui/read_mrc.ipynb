```python
# %%
import mrcfile
import numpy as np
import matplotlib.pyplot as plt
from glob import glob


def read_mrc(file_path):
    with mrcfile.open(file_path) as mrc:
        data = mrc.data
    return data
```

```python
n = 99

plt.figure(figsize=(15, 8))

data = read_mrc(glob(f"/home/tomo/kaggle/polnet/output/241130/tomogram_{n}/tomos/tomo_rec_0_snr*.mrc")[0])
plt.subplot(1, 2, 1)
plt.imshow(data[data.shape[0]//2], cmap="gray")

data = read_mrc(f"/home/tomo/kaggle/polnet/output/241130/tomogram_{n}/tomos/tomo_lbls_0.mrc")
# print(np.unique(data))
plt.subplot(1, 2, 2)
plt.imshow(data[data.shape[0]//2])
plt.show()
```
